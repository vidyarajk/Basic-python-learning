```python
lambda1=lambda x:x**2
def applyOnList(l1,lambda1):
    l2=[]
    for i in l1:
        l2.append(lambda1(i))
    return l2
print(applyOnList([1,2,3,4,5],lambda1))

##OR
print(applyOnList([3,4,5,6],lambda x:x**3))
```
