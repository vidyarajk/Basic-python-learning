##### Write a program to get a list and return each and every element in a positve/abs value

```python
def absolute(x):
    return abs(x)
absolute(-4)
```

```python
def listOfAbs(l1,absolute):
    l2=[]
    for i in l1:
        l2.append(absolute(i))
    return l2
print(listOfAbs([-4,-2,-8,-9,4],absolute))
```

```python
def positive(x):
    return x>=0
```

```python
def FilterMylst(l1,fun):
    l1modified=[]
    for i in l1:
        if (fun(i)):
            l1modified.append(i)
    return l1modified
```

```python
print(FilterMylst([-1,-2,-3,4,100,2],positive))
```

#### Return function from another function

```python
def greet(name):
    return f"Namaste {name}"
def hello():
    return greet("all")
print(greet("all"))
```

```python
import math
```

```python
def factorial():
    return math.factorial
def FindFact(num):
    f=factorial()
    return f(num)
print(FindFact(5))

```
